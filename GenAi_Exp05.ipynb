```python
# ===============================
# Tool-Using Intelligent Agent

import re
import math

# ----------------------------
# Step 1: Define Tools
# ----------------------------



def calculator(expression):
    """Advanced Calculator Tool (Safer)"""
    try:
        allowed_names = {
            "sqrt": math.sqrt,
            "pow": pow,
            "abs": abs,
            "round": round
        }

        result = eval(expression, {"__builtins__": None}, allowed_names)
        return str(result)

    except Exception as e:
        return "Error in calculation"

def knowledge_base(question):
    """Simple Knowledge Tool (Simulated LLM Knowledge)"""
    data = {
        "dbms": "DBMS stands for Database Management System. It manages databases efficiently.",
        "java": "Java is an object-oriented programming language developed by Sun Microsystems.",
        "ai": "Artificial Intelligence is the simulation of human intelligence in machines."
    }

    for key in data:
        if key in question.lower():
            return data[key]

    return "I don't have information about that."

# ----------------------------
# Step 2: Agent Logic
# ----------------------------

def agent(user_input):

    print("\n Thought: Understanding the user query...")

    # Better math detection
    if any(op in user_input for op in "+-*/%^"):
        print(" Action: Using Calculator Tool")

        expression = user_input.replace(" ", "")
        result = calculator(expression)

        print(" Observation:", result)
        return f"Final Answer: {result}"

    else:
        print(" Action: Using Knowledge Base Tool")

        result = knowledge_base(user_input)
        print(" Observation:", result)

        return f"Final Answer: {result}"

# ----------------------------
# Step 3: Run Agent
# ----------------------------

while True:
    user_input = input("\nAsk something (type 'exit' to stop): ")

    if user_input.lower() in ["exit", "stop"]:
        break

    answer = agent(user_input)
    print(answer)
```
