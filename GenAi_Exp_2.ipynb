```python
def chain_of_thought(prompt):
    system_message = (
        "You are an advanced reasoning assistant. "
        "Always think step-by-step and explain your reasoning clearly."
    )

    messages = [
        {"role": "system", "content": system_message},
        {"role": "user", "content": prompt}
    ]

    response = client.responses.create(
        model="gpt-4o-mini",
        input=messages,
        temperature=0.3
    )

    return response.output_text
```

```python
question = "A train travels 60 km in 1.5 hours. What is its speed?"
print(chain_of_thought(question))
```

```python
def few_shot_classification(prompt, examples):
    system_prompt = "You are a text classification assistant."

    messages = [{"role": "system", "content": system_prompt}]

    # Add few-shot examples
    for ex in examples:
        messages.append({"role": "user", "content": ex["input"]})
        messages.append({"role": "assistant", "content": ex["output"]})

    # Add final query
    messages.append({"role": "user", "content": prompt})

    response = client.responses.create(
        model="gpt-4o-mini",
        input=messages,
        temperature=0.2
    )

    return response.output_text
```

```python
examples = [
    {"input": "I love this product!", "output": "Sentiment: Positive"},
    {"input": "This is the worst movie ever.", "output": "Sentiment: Negative"},
    {"input": "The food was okay.", "output": "Sentiment: Neutral"}
]
```

```python
result = few_shot_classification("The food was amazing!", examples)
print(result)
```

```python
result = few_shot_classification("The food was amazing But prize is to high!", examples)
print(result)
```
